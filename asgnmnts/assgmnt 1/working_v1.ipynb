```python
from gdeltdoc import GdeltDoc, Filters
import time
import pandas as pd

gd = GdeltDoc()

months = [
    ("2020-01-01", "2020-01-31"),
    ("2020-02-01", "2020-02-28"),
    ("2020-03-01", "2020-03-31"),
]

all_articles = []

for start, end in months:
    try:
        f = Filters(
            keyword="Tesla ESG",
            start_date=start,
            end_date=end
        )
        articles = gd.article_search(f)
        print(f"{start} to {end}: {len(articles)} articles found")
        all_articles.append(articles)
        time.sleep(5)  # wait 5 seconds between each call
    except Exception as e:
        print(f"Error for {start} to {end}: {e}")
        time.sleep(10)  # wait longer if error occurs

# combine all results
df = pd.concat(all_articles, ignore_index=True)
print(f"\nTotal articles: {len(df)}")
print(df.head())
```
